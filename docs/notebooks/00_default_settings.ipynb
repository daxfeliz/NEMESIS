# 00 — NEMESIS default pipeline settings

This tutorial explains the settings returned by `default_NEMESIS_pipeline_settings()`. Start here before running [Notebook 01](01_nemesis_leovetter.ipynb).

**Source of truth:** The tables below are generated directly from the effective dictionary in `src/nemesis_tess/pipeline.py`. This document describes the **software defaults**, which are not necessarily the science configuration used for the NEMESIS II survey. Values are displayed without executing the pipeline.

There are **85 effective settings** in the current source. You can execute the optional code cells in a Python environment with NEMESIS installed, but the Read the Docs build will not execute them.


## 1. Retrieve the current defaults

This function returns a Python dictionary. Access a value by its key, and copy the dictionary before changing options. The example imports the installed NEMESIS package; it does not download TESS data.


```python
from nemesis_tess import default_NEMESIS_pipeline_settings

settings = default_NEMESIS_pipeline_settings()
print(f'{len(settings)} settings loaded')
print('Transit search enabled:', settings['do_transit_search'])
print('Minimum period (days):', settings['minP'])
print('Required transit count:', settings['N_transits'])
```

## 4. Override settings for a single run

Changing a dictionary does not change the global defaults. Pass the modified settings to `full_pipeline(..., settings=settings)`. The example below enables verbose logs, retains FITS products, and chooses TIC rather than Gaia for plot annotations. This is useful where remote Gaia/VizieR queries are unavailable.


```python
settings = default_NEMESIS_pipeline_settings()
settings.update({
    'verbose': True,
    'keep_FITS': True,
    'plot_catalog_source': 'TIC',
})
for key in ('verbose', 'keep_FITS', 'plot_catalog_source'):
    print(f'{key} = {settings[key]!r}')

# Run only when you intend to download and process TESS data:
# from nemesis_tess import full_pipeline
# full_pipeline(ID=103633434, path='output/', cadence='30 minute',
#               Sector=20, settings=settings)
```

## 5. Compare your configuration with the defaults

When debugging a run, print only values that differ from the defaults. This is safer than copying the whole settings dictionary into documentation and forgetting to update it.


```python
defaults = default_NEMESIS_pipeline_settings()
changed = {k: (defaults[k], settings[k]) for k in defaults
           if settings[k] != defaults[k]}
for key, (before, after) in changed.items():
    print(f'{key}: {before!r} -> {after!r}')
assert set(settings) == set(defaults)
```
