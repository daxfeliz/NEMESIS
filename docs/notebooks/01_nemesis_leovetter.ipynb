# NEMESIS: a blind BLS/TLS search with integrated LEO-Vetter

**Demonstration target:** TIC 103633434 (TOI-1235), TESS Sector 20, 30-minute cadence.
Requires internet access (MAST/TESScut/Gaia). This is the tested `full_pipeline` entrypoint,
not a standalone LEO-Vetter invocation. The report's sixth page runs LEO-Vetter automatically.

**Important:** Retain FITS and intermediate aperture files so notebook 03 can later
reproduce the exact aperture geometry used for statistical validation.
The folder below is the shared output root for notebooks 01–03.


```python
from pathlib import Path
from nemesis_tess import full_pipeline, default_NEMESIS_pipeline_settings

TIC_ID = 103633434
SECTOR = 20
OUTPUT = Path.home() / 'nemesis_tutorials' / 'toi1235_s20'
OUTPUT.mkdir(parents=True, exist_ok=True)

settings = default_NEMESIS_pipeline_settings()
settings['keep_imagedata'] = True  # preserve *_pix_mask.pkl for TRICERATOPS
settings['keep_extra_files'] = True
settings['keep_FITS'] = True       # preserve the matching TESScut FITS
settings['keep_extra_transit_files'] = True  # retain BLS/TLS report files

hdu, lc, transit_results, run_settings = full_pipeline(
    ID=TIC_ID,
    path=str(OUTPUT) + '/',
    Sector=SECTOR,
    cadence='30 minute',
    settings=settings,
)

assert hdu is not None, 'No TESScut HDU returned'
assert lc is not None and not lc.empty, 'No light curve produced'
assert transit_results is not None and len(transit_results) == 6
print('Light curve cadences:', len(lc))
print('BLS/TLS results:', [type(x).__name__ for x in transit_results])
```

## Inspect the light curve and transit reports

The six returned entries contain the BLS power spectrum, model and parameters,
followed by the corresponding TLS entries. For table columns and LEO-Vetter
diagnostics use the saved CSVs, not guesses about column names. A successful
run can have no significant transit; interpret the search results scientifically.


```python
import pandas as pd
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 3))
ax.plot(lc['Time'], lc['Detrended Flux'], '.', ms=2)
ax.set(xlabel='Time (BTJD)', ylabel='Detrended relative flux',
       title=f'TIC {TIC_ID} / Sector {SECTOR}')
plt.show()

tls_root = Path(run_settings['saveReportpath_TLS'])
bestfit_files = sorted(tls_root.rglob(
    f'TIC_{TIC_ID}_Sector_{SECTOR:04d}_TLS_bestfit.csv'
))
assert len(bestfit_files) == 1, f'Expected 1 TLS bestfit; found: {bestfit_files}'
bestfit = pd.read_csv(bestfit_files[0])
print('TLS bestfit:', bestfit_files[0],'\n')
print('LEO-Vetter fields:', [c for c in bestfit.columns if c.startswith('LEOVetter_')],'\n')
display(bestfit.head())
```

**Acceptance criteria:** a nonempty light curve, both search branches,
best-fit CSVs and the PDF vetting reports. If the `LEOVetter_*` columns are
missing, inspect page-six generation and errors rather than silently
claiming LEO-Vetter passed. Keep the output folder for notebooks 02–03.
